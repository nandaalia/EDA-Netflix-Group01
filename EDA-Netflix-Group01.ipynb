```python
# Import library dasar
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Baca data
df = pd.read_csv('netflix_titles.csv')

# Lihat 5 data teratas
df.head()

df.info()

df.describe()

sns.countplot(data=df, x='type')
plt.title('Jumlah Konten di Netflix (Movie vs TV Show)')
plt.show()

plt.figure(figsize=(12,6))
df['release_year'].value_counts().sort_index().plot(kind='bar')
plt.title('Jumlah Konten Netflix berdasarkan Tahun Rilis')
plt.xlabel('Tahun')
plt.ylabel('Jumlah Konten')
plt.show()


# Isi nilai kosong dengan nilai default
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Not specified')
df['country'] = df['country'].fillna('Unknown')

# Hapus baris yang punya missing values di kolom yang penting
df = df.dropna(subset=['date_added', 'rating', 'duration'])

df.isnull().sum()















```
